```python
import numpy as np
import pandas as pd
import statistics
```

- What is the standard deviation for the numbers: 75, 83, 96, 100, 121 and 125?

```python
data = [75, 83, 96, 100, 121, 125]


# Numpy 

samp_std_np = np.std(data, ddof=1).round(2)
popu_std_np = np.std(data).round(2)

print(f'Np.Popu-STD = {popu_std_np}')
print(f'Np.Samp-STD = {samp_std_np}\n')


# Pandas

samp_std_pd = round(pd.Series(data).std(), 2)
popu_std_pd = round(pd.Series(data).std(ddof=0), 2)

print(f'Pd.Popu-STD = {popu_std_pd}')
print(f'Pd.Samp-STD = {samp_std_pd}')
```

- What is the range ?



```python
temp = [40, 45, 50, 45, 35, 25, 10]


# Numpy

np_range = np.ptp(temp)
print(f'Np_Range= {np_range}')


# Pandas

temp_series = pd.Series(temp)
pd_range = temp_series.max() - temp_series.min()

print(f'Pd_Range= {pd_range}')
```

- Nine friends each guessed the number of marbles in a jar.

- Emma rolled a die a number of times and recorded her results in a bar graph, as follows:

```python
scores = [1, 2, 3, 4, 5, 6]
frequencies = [7, 8, 6, 4, 7, 8]


# Numpy 

data = np.repeat(scores, frequencies)

samp_var_np = np.var(data, ddof=1).round(2)
popu_var_np = np.var(data).round(2)

print(f'Np.Popu-var = {popu_var_np}')
print(f'Np.Samp-var = {samp_var_np}\n')


# Pandas

samp_var_pd = round(pd.Series(data).var(), 2)
popu_var_pd = round(pd.Series(data).var(ddof=0), 2)

print(f'Pd.Popu-var = {popu_var_pd}')
print(f'Pd.Samp-var = {samp_var_pd}')
```

- Calculate the Pearson correlation coefficient for the following data:



```python
x = [12, 8, 5, 3, 2, 0]
y = [1, 7, 4, 6, 4, 2]

# Numpy


corr_np = np.corrcoef(x, y).round(2)
print(f'Np_Correlation = {corr_np[0, 1]}') 


# Pandas

corr_pd = pd.Series(x).corr(pd.Series(y)).round(2)
print(f'Pd_Correlation = {corr_pd}')
```
