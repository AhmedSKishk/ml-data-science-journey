- Write a function that returns the lesser of two given numbers if both numbers are even, but returns the greater if one or both numbers are odd</p>

    - (2,5) → 5

```python
# n = (2, 4), (2, 5)
n1 = int(input('Enter number 1:'))
n2 = int(input('Enter number 2:'))

def lesser_of_two_numbers(n1, n2):

    # use if and modulus(%) to check even
    if n1 %2 == 0 and n2 %2 ==0:
        print(min(n1, n2))

    else:
        print(max(n1, n2))

lesser_of_two_numbers(n1, n2)
```

- Given a sentence, return a sentence with the words reversed</p>
    - ('I am home') → 'home am I'

```python
# sentence = ('I am home') → 'home am I'
sentc = input('Enter a sentence:')

def reverse_sentc(sentc):
    
    sentc = sentc.split()
    sentc.reverse()
    sentc = ' '.join(sentc)
    return sentc

reverse_sentc(sentc)
```

```python
# sentence = ('I am home') → 'home am I'
sentc = input('Enter a sentence:')

def reverse_sentc(sentc):
    
    sentc = sentc.split()
    sentc = sentc[::-1]
    sentc = ' '.join(sentc)
    return sentc

reverse_sentc(sentc)
```

- Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.

```python
month = input('Enter a month name:')

def check_season(month):

    if month in ['december', 'january', 'february']:
        return 'Winter'
    elif month in ['march', 'april', 'may']:
        return 'Spring'
    elif month in ['june', 'july', 'august']:
        return 'Summer'
    elif month in ['september', 'october', 'november']:
        return 'Autumn'
    else:
        return 'Invalid month'
```

- Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.
    - Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    - Expected Output :
        - No. of Upper case characters : 4
        - No. of Lower case Characters : 33

```python
def count(text):

    count_upper = 0
    count_lower = 0
    for char in text:
    
        if char.isupper():
            count_upper = count_upper + 1
        elif char.islower():
            count_lower = count_lower + 1
    
    print("No. of Upper case characters :", count_upper)
    print("No. of Lower case Characters :", count_lower)
```

```python
count('Hello Mr. Rogers, how are you this fine Tuesday?')
```

- Write a Python function that checks whether a passed in string is palindrome or not.
    - Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g., `madam` or `nurses run`.

```python
def is_palindrome(text):
    # Remove spaces and make lowercase to properly check phrases like 'nurses run' | 'madam'
    clean_text = text.replace(' ', '').lower()
    return clean_text == clean_text[::-1]
```

- Write a Python function to multiply all the numbers in a list.</p>
    - Sample List : [1, 2, 3, -4]
    - Expected Output : -24

```python
def multiply_list(numbers):

    result = 1  
    for num in numbers:
        result *= num
    return result

sample_list = [1, 2, 3, -4]
output = multiply_list(sample_list)

print(f"Sample List: {sample_list}")
print(f"Expected Output: {output}")
```
