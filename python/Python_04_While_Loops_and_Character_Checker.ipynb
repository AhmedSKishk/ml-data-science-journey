```python
# name = 'ahmed'

name = input('Enter your name:')
idx = 0

while idx < len(name):

    if name[idx].lower() == 'e':
        break

    print(name[idx])
    idx += 1
```

- Skip the Letter 'e' While Printing Characters of a String

```python
# name = 'ahmed'

name = input('Enter your name: ')
idx = 0

while idx < len(name):
    if name[idx].lower() == 'e':
        idx += 1
        continue
    
    print(name[idx])
    idx += 1
```

- Check if a String Contains 'e' Using a While-Else Loop
- Search for the Letter 'e' Case-Insensitively Using a While-Else Loop

```python
# name = 'ahmed'

name = input('Enter your name: ')
idx = 0

while idx < len(name):
    if name[idx].lower() == 'e':
        print('Letter `e` was found')
        break

    print(name[idx])
    idx += 1
else:
    print('Letter `e` was not found')
```

### Take string input value of one character, then check the following : 
- If character is aplphabetic --> print that it is alphabetic
- If character is Numeric --> print that it is a number.
- IF character is a special character --> print that it is a special character.

Using Nested IF Condition
- If alpbhabetic --> Check whether it is lowercase or uppercase.
- If numeric --> Check whether it is positive, negative, zero.

### Make input takes email, then loop on each character in the email and apply the previous conditions
### Make the user have the option to insert more than one email repeateadly.

```python
while True:

    email = input('Enter your email: ')

    # email = Ahmed5@gmail

    for char in email:

        if char.isalpha():
            # print(f'Character `{char}` is alphabetic')

            if char.islower():
                print((f'Character `{char}` is alphabetic and lowercase'))

            else:
                print((f'Character `{char}` is alphabetic and uppercase'))


        elif  char.replace('-', '').isnumeric():
            # print(f'Character `{char}` is number')

            char = int(char)

            if char > 0:
                print(f'Character `{char}` is number and postive')

            elif char < 0:
                print(f'Character `{char}` is number and negative')

            else:
                print('Character is zero')


        else:
            print(f'Character `{char}` is a special character')

    proceed = input('Do you want to proceed ? [Y/N]').upper()
    
    if proceed == 'N':
        print('Thank you for using our program')
        break
```
