- Write a Python program to Check if a Triangle is (Equilateral متساوي الأضلاع), (Isosceles متساوي الساقين), or (Scalene مختلف الأضلاع).
- Take `a`, `b`, `c` as inputs from the user

```python
# a = 12 , b = 12 , c = 12

a = float(input('Enter side a:'))
b = float(input('Enter side b:'))
c = float(input('Enter side c:'))

if a == b == c :
    print('Equilateral متساوي الأضلاع')

elif a == b or b == c or c == a:
    print('Isosceles متساوي الساقين') 

else:
    print('Scalene مختلف الأضلاع')
```

Take string input value of one character, then check the following : 
- If character is aplphabetic --> print that it is alphabetic
- If character is Numeric --> print that it is a number.
- IF character is a special character --> print that it is a special character.

Using Nested IF Condition
- If alpbhabetic --> Check whether it is lowercase or uppercase.
- If numeric --> Check whether it is positive, negative, zero.

```python
# Single character = @

char = input('Enter a single character:')

if char.isalpha():
    print(f'`{char}` is alphabetic')

    if char.islower():
        print(f'`{char}` is alphabetic and lowercase')
    else:
        print(f'`{char}` is alphabetic and uppercase')
            

elif char.lstrip('-').isnumeric():
    print(f'`{char}` is number')

    num=int(char)

    if num > 0:
        print(f'`{char}` is number and positive')

    elif num < 0:
        print(f'`{char}` is number and negative')

    else:
        print(f'`{char}` is number and equal zero')

else:
    print(f'`{char}` is special character')
        
```
