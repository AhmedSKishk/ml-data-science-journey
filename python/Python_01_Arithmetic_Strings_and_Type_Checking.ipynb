Basic Arithmetic Operations
- Write a program that takes two integers with variables names `a` and `b`, and prints their sum, difference, multiplication, exponent, division, floor division and modulus.

```python
a = int(input('Enter the first integer:'))
b = int(input('Enter the second integer:'))

print('sum =', a + b) 
print('difference =', a - b)    
print('multiplication = ', a * b) 
print('exponent = ', a ** b)   
print('division = ', a / b)
print('floor division = ', a // b)   
print('modulus = ', a % b)    
```

Basic Arithmetic Operations
- Given the radius of a circle, calculate circumference (Search for the rule if you don't know it)

```python
radius = 5

import math

circumference = 2 * math.pi * radius
print('Circumference_of_circle = ', circumference)
```

- Given the width and height of a rectangle, calculate the perimeter (Search for the rule if you don't know it)

```python
length = 5
width = 10

perimeter = 2 * (length + width)
print('Perimeter of rectangle =', perimeter)
```

String Indexing and Slicing
- Write a program that takes a string `s` and prints :
* first character
* last character
* Substring from (e) character to (w) character.

```python
s = 'helloworld'

print(s[0])
print(s[9])
print(s[1:6])
```
