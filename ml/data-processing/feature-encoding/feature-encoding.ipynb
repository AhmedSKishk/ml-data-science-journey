# Feature Encoding

```python
import numpy as np
import pandas as pd
import plotly.express as px
from sklearn.preprocessing import OneHotEncoder
from category_encoders import BinaryEncoder
from sklearn.preprocessing import OrdinalEncoder

pd.set_option('display.max_columns', None)
```

### 1. One Hot Encoding

Use when the categories have no order and there are few of them.

```python
df = px.data.tips()
df
```

- 1.1) First method

```python
pd.get_dummies(df['sex'], drop_first= True)
```

```python
pd.get_dummies(df['day'])
```

- 1.2) Second method(Sklearn)

```python
ohe = OneHotEncoder()
ohe.fit_transform(df[['sex']])
```

```python
ohe = OneHotEncoder(sparse_output= False, drop= 'first')
ohe_array = ohe.fit_transform(df[['sex']])
ohe_array
```

```python
ohe = OneHotEncoder(sparse_output= False, drop= 'first')
df['sex'] = ohe.fit_transform(df[['sex']])
df.head(1)
```

```python
ohe = OneHotEncoder(sparse_output= False, drop= 'first')
ohe_array = ohe.fit_transform(df[ ['smoker', 'day', 'time'] ])
cols_name = ohe.get_feature_names_out()
ohe_df = pd.DataFrame(ohe_array, columns= cols_name)
ohe_df.head(1)
```

```python
df = pd.concat([df, ohe_df], axis= 1).drop(['smoker', 'day', 'time'], axis= 1)
df.head(1)
```

### 2. Binary Encoder

Use when a column has many categories. One-hot would create 288 columns here, binary needs 19.

```python
df = px.data.gapminder()
df
```

```python
be = BinaryEncoder()
be_df = be.fit_transform(df[['country', 'continent', 'iso_alpha']])
be_df.head(1)
```

```python
df = pd.concat([df.drop(columns=['country', 'continent', 'iso_alpha']), be_df], axis=1)
df.head(1)
```

### 3. Ordinal Encoding

Use when the categories have a natural order (here XS < S < M < L < XL < XXL).

```python
PATH = r'dataset/t-shirts.csv'
df = pd.read_csv(PATH)
df
```

- 3.1) First method

```python
display(df['Size'].unique())

# Sizes from the smallest to the largest
size_order = ['XS', 'S', 'M', 'L', 'XL', 'XXL']
mapping_dict = {size: i for i, size in enumerate(size_order)}
mapping_dict
```

```python
df['Size'].map(mapping_dict)
```

- 3.2) Second method(Sklearn)

```python
ordinal_enc = OrdinalEncoder(categories= [size_order])
df['Size'] = ordinal_enc.fit_transform(df[ ['Size'] ])
df.head(1)
```

```python
ohe = OneHotEncoder(sparse_output= False, drop= 'first')
ohe_array = ohe.fit_transform(df[ ['Brand', 'Color', 'Design', 'Delivery status'] ])
cols_name = ohe.get_feature_names_out()
ohe_df = pd.DataFrame(ohe_array, columns= cols_name)
ohe_df.head(1)
```

`Order number` is an ID (500 different values in 500 rows) and `Date ordered` is text, so both are dropped instead of encoded.

```python
df = pd.concat([df.drop(columns=['Brand', 'Color', 'Design', 'Delivery status', 'Order number', 'Date ordered']), ohe_df], axis= 1)
df.head(1)
```
