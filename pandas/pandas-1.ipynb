```python
import pandas as pd
import numpy as np
```

- Write a Pandas program to build DataFrame from the following Data

    * names = ['Ahmad', 'Fatma', 'Mahmoud', 'Anas', 'Omar', 'Sayed', 'Ali', 'Maryam', 'Habiba', 'Alaa']
    * scores = [12.5, 9, 16.5, 17, 9, 20, 14.5, 10, 8, 19]
    * attempts = [1, 3, 2, 3, 2, 3, 1, 1, 2, 1]
    * qualify = ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']

```python
data = { 
    'scores' : [12.5, 9, 16.5, 17, 9, 20, 14.5, 10, 8, 19],
    'attempts' : [1, 3, 2, 1, 2, 1, 3, 2, 1, 2],
    'qualify' : ['yes', 'no', 'yes', 'no', 'no', 'yes', 'yes', 'no', 'no', 'yes']
}

names = ['Ahmad', 'Fatma', 'Mahmoud', 'Anas', 'Omar', 'Sayed', 'Ali', 'Maryam', 'Habiba', 'Alaa']

df = pd.DataFrame(data, index=pd.Index(names, name='Name Of Student'))
df
```

- Write a Pandas program to select the 'name' and 'score' columns from the Previous DataFrame

```python
df_1 = df.reset_index()[['Name Of Student', 'scores']]
print(df_1, '\n')

df_2 = df.reset_index()
print(df_2.loc[:, ['Name Of Student', 'scores']])
    
```

- Write a pandas program to add 5 to `scores` column

```python
df['updated_scores'] = df['scores'] + 5
df
```

- Write a Pandas program to show students with score > 20.

```python
df[df['updated_scores'] > 20]
```

- Write a Pandas program to show 'qualified' students.

```python
qualified_students = (df[df['qualify'] == 'yes'])
qualified_students
```

- Write a pandas program to remove the last row in the dataframe.

```python
df.drop(df.index[-1], inplace=True)
df
```

- Write a pandas program to find the minimum, maximum, and range of scores.

```python
min_scores = df['scores'].min()
max_scores = df['scores'].max()
range_scores = max_scores - min_scores

print(f'Min Scores= {min_scores}\nMax Scores {max_scores}\nRange Scores={range_scores}')
```
